```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
omega_0 = 0.9 * 400 * 5 * 10e-9 # keep the volume

# x: radial distance in units of lambda * f-number
x = np.linspace(0, 5e-5, 500)
# Gaussian approximation of Airy disk
I = np.exp(-2 * x**2 / omega_0**2)
I2 = np.exp(-2 * x**2 / omega_0**2) / 2.512 ** 2
I3 = np.minimum(np.exp(-2 * x**2 / omega_0**2) * 100, 1)

x_mag0_size = omega_0 * np.sqrt(-0.5 * np.log(1/255))
plt.axvline(x=x_mag0_size, color='blue', linestyle='--', label='mag 0 size')

x_mag0_size = omega_0 * np.sqrt(-0.5 * np.log(1/255 * 2.512 ** 2))
plt.axvline(x=x_mag0_size, color='orange', linestyle='--', label='mag +2 size')

x_mag0_size = omega_0 * np.sqrt(-0.5 * np.log(1/255 / 100))
plt.axvline(x=x_mag0_size, color='green', linestyle='--', label='mag -5 size')

plt.plot(x, I, label='mag 0', color='blue')
plt.plot(x, I2, label='mag +2', color='orange')
plt.plot(x, I3, label='mag -5', color='green')
plt.xlabel('Radial distance (units of λ·f-number)')
plt.ylabel('Relative radiance')
plt.title('Gaussian Approximation of Airy Disk')
plt.grid(True)
plt.legend()
plt.show()

# plt.savefig("gaussian_airy_disk.png", dpi=300)
# plt.close()  # Close the current figure to avoid saving a blank one
```

```python

# Analytical solution for x where radiance equals target_radiance
target_radiance = 1/255
x_at_target_analytical = omega_0 * np.sqrt(-0.5 * np.log(target_radiance / (100))) / 4.63e-6
print(f"Analytical x where radiance equals {target_radiance}: {x_at_target_analytical}")
```
