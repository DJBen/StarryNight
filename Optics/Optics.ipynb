```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
omega_0 = 1.22 * 400 * 5 * 10e-9 # keep the volume

# x: radial distance in units of lambda * f-number
x = np.linspace(0, 10e-5, 500)
# Gaussian approximation of Airy disk
I = np.exp(-2 * x**2 / omega_0**2) * 100

plt.plot(x, I, label='Gaussian Approximation')
plt.xlabel('Radial distance (units of λ·f-number)')
plt.ylabel('Relative radiance')
plt.title('Gaussian Approximation of Airy Disk')
plt.grid(True)
plt.legend()
plt.show()

# Analytical solution for x where radiance equals target_radiance
target_radiance = 1/255
x_at_target_analytical = omega_0 * np.sqrt(-0.5 * np.log(target_radiance / (100))) / 4.63e-6
print(f"Analytical x where radiance equals {target_radiance}: {x_at_target_analytical}")
```
